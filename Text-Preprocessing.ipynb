Remove HTML Tags

```python
import re
def remove_html_tags(text):
    pattern = re.compile('<.*?>')
    return pattern.sub(r'', text)
```

```python
text = "<html><body><p> Movie 1</p><p> Actor - Aamir Khan</p><title> Home Page - 1 </title></body></html>"
```

```python
remove_html_tags(text)
```

```python
df['review'] = df['review'].apply(remove_html_tags)
```

Remove URL

```python
def remove_url(text):
    pattern = re.compile(r'https?://\S+|www\.\S+')
    return pattern.sub(r'', text)
```

```python
text1 = 'Check out my youtube https://www.youtube.com/dswithbappy dswithbappy'
text2 = 'Check out my linkedin https://www.linkedin.com/in/boktiarahmed73/'
text3 = 'Google search here www.google.com'
text4 = 'For data click https://www.kaggle.com'
```

Punctuation Handling

```python
import string, time
```

```python
string.punctuation
```

```python
exclude = string.punctuation
```

```python
exclude
```

```python
def remove_punc(text):
    for char in exclude:
        text = text.replace(char, '')
    return text
```

```python
text = 'string . with. Punctuation?'
```

```python
start = time.time()
print(remove_punc(text))
time1 = time.time() - start
print(time1*50000)
```

```python
def remove_punc1(text):
    return text.translate(str.maketrans('','',exclude))
```

```python
start = time.time()
remove_punc1(text)
time2 = time.time() - start
print(time2*50000)
```

Chat Conversion Handle

```python
chat_words = {
    'AFAIK':'As Far As I Know',
    'AFK':'Away From Keyboard',
    'ASAP':'As Soon As Possible',
    'ATK':'At The Keyboard',
    'ATM':'At The Moment'
}
chat_words
```

```python
def chat_conversion(text):
    new_text = []
    for w in text.split():
        if w.upper() in chat_words:
            new_text.append(chat_words[w.upper()])
        else:
            new_text.append(w)
    return " ".join(new_text)
```

```python
chat_conversion('Do this Work ASAP')
```

StopWords

```python
from nltk.corpus import stopwords
import nltk
nltk.download('stopwords')
```

```python
stopwords.words('english')
```

```python
len(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    new_text = []
    for word in text.split():
        if word in stopwords.words('english'):
            new_text.append('')
        else:
            new_text.append(word)
    x = new_text[:]
    new_text.clear()
    return " ".join(x)
```

```python
remove_stopwords('probably my all time favorite move, a story of selflessnes, sacrifice and emotion')
```

```python
df['review'] = df['review'].apply(remove_stopwords)
```

Emoji Removal

```python
import re
```

```python
def remove_emoji(text):
    emoji_pattern = re.compile("["
                           u"\U0001F600-\U0001F64F"  # emoticons
                           u"\U0001F300-\U0001F5FF"  # symbols & pictographs
                           u"\U0001F680-\U0001F6FF"  # transport & map symbols
                           u"\U0001F1E0-\U0001F1FF"  # flags (iOS)
                           u"\U00002702-\U0001F251"
                           "]+", flags = re.UNICODE)
    return emoji_pattern.sub(r'', text)
```

```python
remove_emoji("Loved the movie. It was 😘😘")
```
