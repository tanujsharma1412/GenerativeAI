Hugging Face Tasks

```python
from transformers import pipeline

## ------------------- NLP Tasks ------------------



'''
1. Text Classification : Assigning a category to a piece of text.
Sentiment Analysis
Topic Classification
Spam Detection
'''

classifier = pipeline("text-classification")

'''
2. Token classification: Assigning lables to individual tokens in a sequence.
Named Entity Recognition (NER)
Part-of-Speech Tagging
'''

token_classifier = pipeline("token-classification")


'''
3. Question Answering : Extracting an answer from a given context based on question.
'''

question_answerer = pipeline("question-answering")


'''
4. Text Generation : Generating text based on a given prompt.
language Modeling
Story Generation
'''

text_generator = pipeline("text-generation")


'''
5. Summarization : Generating a short summary of a long text.
'''

summarizer = pipeline("summarization")


'''
Translation : Translating text from one language to another.
'''

translator = pipeline("translation", model="Helsinki-NLP/opus-mt-en-fr")



'''
6. Text2Text Generation: General-purpose text transformation, including summarization and translation
'''

text2text_generator = pipeline("text2text-generation")

'''
7. Fill-Mask : Predicting the masked token in a sequence.
'''

fill_mask = pipeline("fill-mask")


'''
8. Feature Extraction: Extracting hidden states or features from text.
'''

feature_extractor = pipeline("feature-extraction")

'''
9. Sentence Similarrity: Measuring the similarity between two sentences.
'''

sentence_similarity = pipeline("sentence-similarity")


## ----------------- Computer Vision Tasks -----------------------


'''
1. Image Classification: Classifying the main contect of an image.
'''

image_classifier = pipeline("image-classification")


'''
2. Object Detection: Locating and identifying objects in an image.
'''

object_detector = pipeline("object-detection")


'''
3. Image Segmentation: Assigning a label to each pixel in an image.
'''

image_segmenter = pipeline("image-segmentation")


'''
4. Image Generation: Creating new images based on a given prompt.
'''

image_generator = pipeline("image-generation")


## ------------------  Speech Processing Tasks --------------------

'''
1. Utomatic Speech Recognition (ASR): Converting speech to text.
'''

asr = pipeline("automatic-speech-recognition")


'''
2. Speech Translation : Translating speech from one language to another.
'''

speech_translator = pipeline("speech-to-speech")


'''
3. Audio Classification : Classifying audio signals into predefind categories.
'''

audio_classifier = pipeline("audio-classification")


'''
4. Audio Transcription : Converting spoken language to text.
'''

audio_transcription = pipeline("audio-transcription")


## ----------------------- MultiModel Tasks ---------------------------------

'''
1. Image Captioning : Generating a textual description of an image.
'''

image_captioner = pipeline("image-captioning")


'''
2. Visual Question Answering : Answering questions about images.
'''

visual_question_answerer = pipeline("visual-question-answering")


## -------------------- Other Tasks -----------------------------------

'''
1. Table Question Answering : Answering questions about tabular data.
'''

table_question_answerer = pipeline("table-question-answering")


'''
2. Document Question Answering : Answering questions about documents.
'''

document_qa = pipeline("document-question-answering")

'''
3. Time Series Forecasting : Predicting future values in time series data.
'''

time_series_forecaster = pipeline("time-series-forecasting")


```

Batch Senteniment Analysis

```python
classifier = pipeline(task = "sentiment-analysis")

task_list = ["I really like Autocoders, best models for Anomaly Detection",
             "I am not suere if we CAN acutally Evaluate LLMs.",
             "PassiveAgressive is the name of a Linear Regression Model that so many people do not know.",
             "I hate long meetings"]

classifier(task_list)
```

```python
# Use a pipeline as a high-level Helper

from transformers import pipeline

text_generator = pipeline("text-generation", model = "distilbert/distilgpt2")

generated_text = text_generator("Today is a rainy day in London", truncation = True, num_return_sequences = 2)

print("Generated Text : \n", generated_text[0]['generated_text'])

```

Question Answering

```python
from transformers import pipeline

qa_model = pipeline("document-question-answering", model="impira/layoutlm-document-qa")
question = "Where do I live?"
context = "My name is Merve and I live in İstanbul."

qa_model(question = question, context = context)
```

Tokenization

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, DistilBertTokenizer, DistilBertForSequenceClassification
```

```python


model_name1 = "distilbert-base-uncased-finetuned-sst-2-english"
mytokenizer1 = DistilBertTokenizer.from_pretrained(model_name1)
mymodel1 = DistilBertForSequenceClassification.from_pretrained(model_name1)

classifier = pipeline("sentiment-analysis")
res = classifier("I was so not happy with the Barbie Movie")
print(res)


model_name2 = "nlptown/bert-base-multilingual-uncased-sentiment"
mytokenizer2 = DistilBertTokenizer.from_pretrained(model_name2)
mymodel2 = DistilBertForSequenceClassification.from_pretrained(model_name2)

classifier = pipeline("sentiment-analysis", model= mymodel2, tokenizer= mytokenizer2)
res = classifier("I was so not happy with the Barbie Movie")
print(res)


```
